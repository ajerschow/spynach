# SpynDynamica quick start

[SpynDynamica](https://ajerschow.github.io/spynach/spindynamica.html) is a Python port of
[SpinDynamica](https://www.spindynamica.soton.ac.uk), the Mathematica package for spin dynamics by Malcolm Levitt
and co-workers. Spin operators are sums of product operators with symbolic coefficients (SymPy), and a numeric
engine propagates sequences of pulses and delays. This notebook installs it from PyPI and follows a coupled
spin pair through a pulse and a delay, symbolically.

## Product operators of a spin pair

`I(1, "z")` is the z operator of spin 1 (spins are numbered from 1, as in SpinDynamica) and `@` multiplies
operators. The Hamiltonian has the chemical shift of spin 1 and the coupling, both as symbols.

```python
import sympy as sp
from spyndynamica import SpinSystem, commutator

s = SpinSystem(2)
I = s.I
Omega, J, t = sp.symbols("Omega J t", real=True)
H = Omega * I(1, "z") + 2 * sp.pi * J * (I(1, "z") @ I(2, "z"))
commutator(I(1, "x"), I(1, "y"))
```

## A pulse, then free evolution

A 90° pulse about x on spin 1, then evolution under `H` for the time `t`. The result is a sum of product
operators; `amplitude` picks the coefficient of one of them.

```python
rho = I(1, "z").rotate(sp.pi / 2, "x", spins=[1]).evolve(H, t)
rho
```

```python
antiphase = rho.amplitude(2 * I(1, "x") @ I(2, "z"))
antiphase
```

## The amplitude as a curve

The amplitude is a SymPy expression, so it can be given numbers and plotted: here with a shift of 30 Hz and a
coupling of 7 Hz.

```python
import matplotlib.pyplot as plt
import numpy as np

times = np.linspace(0, 0.3, 600)
curve = sp.lambdify(t, antiphase.subs({Omega: 2 * sp.pi * 30, J: 7}))(times)

plt.plot(times, curve)
plt.xlabel("time (s)")
plt.ylabel("amplitude of 2 I1x I2z")
plt.show()
```
