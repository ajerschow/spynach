# EasySpyn quick start

[EasySpyn](https://ajerschow.github.io/spynach/easyspin.html) is a Python port of the
[EasySpin](https://easyspin.org) EPR simulators, keeping EasySpin's interface: spin systems as dictionaries with
EasySpin's field names, fields in mT, couplings in MHz. This notebook installs it from PyPI and simulates the
cw EPR powder spectrum of a nitroxide.

## A nitroxide powder spectrum

`g` and `A` are the principal values of the tensors (A in MHz), `Nucs` lists the coupled nuclei and `lwpp`
is the peak-to-peak line width in mT. `pepper` returns the field axis `x` and the first-harmonic spectrum `spec`.

```python
import easyspyn as es

nitroxide = {"g": [2.0088, 2.0061, 2.0027], "Nucs": "14N", "A": [16, 16, 86], "lwpp": 0.3}
experiment = {"mwFreq": 9.5, "Range": [330, 350], "nPoints": 1024}

powder = es.pepper(nitroxide, experiment)
```

```python
import matplotlib.pyplot as plt

plt.plot(powder.x, powder.spec)
plt.xlabel("magnetic field (mT)")
plt.ylabel("EPR signal (derivative)")
plt.title("Nitroxide powder spectrum at 9.5 GHz")
plt.show()
```
